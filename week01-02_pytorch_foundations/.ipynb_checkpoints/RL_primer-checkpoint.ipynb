```python
import gymnasium as gym

env = gym.make("CartPole-v1") #builds an environment with cart and pole with the ability to move the cart left or right
observation, info = env.reset(seed=42) # this resets and starts the episode with same parameters such as cart position, pole angle etc 

print("Initial observation:", observation)

for step in range(10):
    action = env.action_space.sample()  # it generates uniform random actions, CartPole has two discrete actions those are 0 (move left) and 1 (move right) 
    observation, reward, terminated, truncated, info = env.step(action)
    # terminated - the the angle of the pole passes a certain limit
    # truncated - the episode fails when a certain time level is time reached.
    print(f"step {step}: action={action}, observation={observation}, reward={reward}, terminated={terminated}")

    if terminated or truncated:
        print("Episode ended, resetting")
        observation, info = env.reset()

env.close() # releases any rendering windows or resources open by the 
```

```python
def run_episode(seed=None):
    env = gym.make("CartPole-v1")

    observation, info = env.reset(seed=seed)

    total_reward = 0
    step_count = 0

    while True:
        # Random action
        action = env.action_space.sample() # it generates uniform random actions, CartPole has two discrete actions those are 0 (move left) and 1 (move right)

        observation, reward, terminated, truncated, info = env.step(action)

        total_reward += reward
        step_count += 1

        if terminated or truncated:
            break

    env.close()

    return step_count, total_reward


# Run multiple episodes
num_episodes = 5

for episode in range(num_episodes):
    # Use different seeds (or set seed=None for completely random episodes)
    steps, total_reward = run_episode(seed=episode)

    print(
        f"Episode {episode + 1}: "
        f"Steps Survived = {steps}, "
        f"Total Reward = {total_reward}"
    )
```
